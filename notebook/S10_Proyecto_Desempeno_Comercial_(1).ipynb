## 🔹 Paso 5: Narrativa con Modelo SCQA + mensaje asincrónico



#### 🖥️ Vista General (Overview)

**S (Situación):**  
Andes Retail Group vende en Perú, Chile y Colombia, en cuatro categorías de producto. En 2024-2025 generó $5,532K en ingresos y $1,942K en ganancia, con un margen del 35.1%.

**C (Complicación):**  
El problema es que el negocio no está creciendo: los ingresos de 2025 cayeron un 6,7% frente a 2024, y eso incluye diciembre, que normalmente es el mes más fuerte del año.

**Q (Pregunta):**   
¿Dónde está generando valor el negocio hoy, y dónde existen palancas claras de mejora sin necesidad de un diagnóstico profundo?

**A (Respuesta):**  
Electrónica y Deportes son las categorías que más venden, y Premium y Estándar son los segmentos que más aportan en dinero — ahí está el valor actual del negocio. Pero la señal que más debería llamar la atención es la caída del 6.7% en ingresos durante 2025. Como el margen es similar en todos los cortes (~35%), el problema no está en la eficiencia de ninguna parte del negocio — está en que se está vendiendo menos volumen. La recomendación inmediata es enfocar los esfuerzos comerciales en recuperar volumen, no en revisar costos o precios; la Vista Detalle identifica dónde concentrar ese esfuerzo primero.

---

#### 🔎 Vista Detalle

**S (Situación):**  
 Tras la caída de ingresos identificada en la Vista General, los datos muestran un patrón claro de estacionalidad: Ropa y Hogar caen notablemente más que el resto de las categorías durante los meses de baja temporada, mientras que Deportes se mantiene relativamente estable todo el año. Al revisar si esa variación responde a diferencias de eficiencia, el margen resulta similar en los tres cortes analizados (país-región, segmento, categoría): entre 34,8% y 35,5%.

**C (Complicación):**  
Si el margen es similar en todos los cortes, entonces algo más está generando diferencias considerables: los ingresos en Verano llegan a ser 3.4 veces los de Invierno, y Colombia genera un tercio menos de ganancia que Perú o Chile. Ese "algo" todavía no se ha nombrado.

**Q (Pregunta):**  
Si no es un tema de eficiencia, ¿qué está causando esas diferencias, y dónde está realmente la oportunidad?

**A (Respuesta):**  
La diferencia está en cuánto se vende, no en qué tan bien se vende. Perú genera 760K en ganancia total, un 62% más que los 470K de Colombia, pero ambos tienen un margen similar (~35%), la diferencia entre ambos está en el volumen de ventas, no en la eficiencia con la que venden. Pasa algo parecido con las estaciones: en Verano se vende 3.4 veces más que en Invierno (2,241,719  contra $653,513). Deportes se mantiene estable todo el año, mientras que Ropa y Hogar dependen mucho de la temporada. En resumen: la eficiencia es similar en todos los cortes analizados, la diferencia entre regiones y estaciones responde principalmente al volumen, no al margen. Ahí es donde probablemente valga la pena poner el foco en el corto plazo.
